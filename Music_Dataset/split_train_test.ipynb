```python
import numpy as np
import pandas as pd
import os
from os import listdir
from os.path import isfile, join
from shutil import copy

data = pd.read_csv('../maestro-v2.0.0/maestro-v2.0.0.csv')

data_files = []
for year in ['2004', '2006', '2008', '2009', '2011', '2013', '2014', '2015', '2017', '2018']:
    data_dir = '../maestro-v2.0.0/' + str(year)
    data_files += [join(str(year), f)for f in listdir(data_dir) if isfile(join(data_dir, f)) if 'midi' in f]

def mkdir(path):
    folder = os.path.exists(path)
    if not folder:                 
        os.makedirs(path)   

def findsplit(name):
    for i in range(len(data)):
        if data['midi_filename'].values[i] == name:
            return data['split'].values[i] 

mkdir('train')   
mkdir('test')      
for item in data_files:
    if findsplit(item) == 'train':
        item = '../maestro-v2.0.0/' + str(item)
        copy(item,'./train')
    elif findsplit(item) == 'test':
        item = '../maestro-v2.0.0/' + str(item)
        copy(item,'./test') 
    
```
